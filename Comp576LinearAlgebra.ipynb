```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 500)
envelope = np.exp(-0.3 * t)
y = envelope * np.sin(2 * np.pi * t)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t, y, color='steelblue', lw=2, label=r'$e^{-0.3t}\sin(2\pi t)$')
ax.plot(t,  envelope, 'r--', lw=1, alpha=0.7, label='envelope')
ax.plot(t, -envelope, 'r--', lw=1, alpha=0.7)
ax.axhline(0, color='gray', lw=0.6)
ax.set_xlabel('time  t')
ax.set_ylabel('amplitude')
ax.set_title('Damped sine wave')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
