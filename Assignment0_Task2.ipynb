Every command from the *Linear Algebra Equivalents* table, run with my own example

### Imports
The two imports the assignment specifies. I alias `scipy.linalg` as `linalg` so the calls match the table (`linalg.inv(a)`, etc.).

```python
import numpy as np
import scipy.linalg as linalg

np.set_printoptions(precision=3, suppress=True)  # tidy output
```

### Example matrices
- `a` — main 4×4 matrix (rows are the digits of *e*, *π*, … just so they're easy to eyeball)
- `b` — a second 4×4 for products / stacking
- `v` — a 1-D vector
- `spd` — a symmetric positive-definite 3×3, needed for the Cholesky example

```python
a = np.array([[2., 7., 1., 8.],
              [3., 1., 4., 1.],
              [5., 9., 2., 6.],
              [4., 2., 8., 3.]])

b = np.array([[1., 0., 2., 0.],
              [0., 3., 0., 1.],
              [4., 0., 1., 0.],
              [0., 2., 0., 5.]])

v = np.array([3., 1., 4., 1.])

spd = np.array([[4., 1., 1.],
                [1., 3., 1.],
                [1., 1., 2.]])

print("a =\n", a)
print("\nb =\n", b)
print("\nv =", v)
print("\nspd =\n", spd)
```

### 1. Size, shape, dimensions

```python
print("ndim(a)      =", np.ndim(a))      # MATLAB: ndims(a)
print("a.ndim       =", a.ndim)
print("size(a)      =", a.size)          # MATLAB: numel(a)  -> total elements
print("shape(a)     =", np.shape(a))     # MATLAB: size(a)
print("a.shape      =", a.shape)
print("a.shape[1]   =", a.shape[1])      # MATLAB: size(a,2)
```

### 2. Constructing / concatenating matrices

```python
# [ 1 2 3; 4 5 6 ]
m = np.array([[1., 2., 3.], [4., 5., 6.]])
print("literal 2x3:\n", m)

# [ a b ; c d ]  -> block matrix
print("\nnp.block:\n", np.block([[np.ones((2,2)), np.zeros((2,2))],
                                    [np.zeros((2,2)), np.eye(2)]]))

# [a b]  (side by side)  and  [a; b]  (stacked)
print("\nhstack (a b):\n", np.hstack((a, b)))
print("\nvstack (a; b):\n", np.vstack((a, b)))
```

### 3. Indexing and slicing
*(indices adapted to the 4×4 where the table assumed a bigger matrix)*

```python
print("a[-1]        (last row, MATLAB a(end,:)) =", a[-1])
print("a[1, 2]      (MATLAB a(2,3))            =", a[1, 2])
print("a[1]         (2nd row, MATLAB a(2,:))   =", a[1])
print("a[0:2]       (first 2 rows)             =\n", a[0:2])
print("a[-2:]       (last 2 rows)              =\n", a[-2:])
print("a[0:3, 1:4]  (rows 1-3, cols 2-4)       =\n", a[0:3, 1:4])
```

```python
# a([1,3,4],[1,3]) style cross-selection -> np.ix_
print("np.ix_ rows{0,2,3} cols{0,2}:\n", a[np.ix_([0, 2, 3], [0, 2])])

# stepped slices
print("\na[0:4:2, :]  (every 2nd row)        =\n", a[0:4:2, :])
print("\na[::2, :]    (every other row)      =\n", a[::2, :])

# reversing rows  (MATLAB flipud / a(end:-1:1,:))
print("\na[::-1, :]   (flipud):\n", a[::-1, :])
print("\na[:, ::-1]   (fliplr):\n", a[:, ::-1])

# a([1:end 1],:)  -> append first row to the end
print("\na[np.r_[:len(a), 0]]:\n", a[np.r_[:len(a), 0]])
```

### 4. Transpose and matrix / element-wise products

```python
print("a.T (transpose):\n", a.T)
print("\na.conj().T (conjugate transpose):\n", a.conj().T)

print("\na @ b (matrix multiply):\n", a @ b)
print("\na * b (element-wise .* ):\n", a * b)
print("\na / b (element-wise ./ ):\n", a / b)
print("\na ** 3 (element-wise .^3):\n", a ** 3)
```

### 5. Comparisons, logic, conditional assignment

```python
print("a > 4:\n", a > 4)
print("\nnonzero(a > 4)  (MATLAB find):", np.nonzero(a > 4))

# a(:, find(v > 2))  -> columns where the vector condition holds
print("\ncolumns of a where v > 2:\n", a[:, np.nonzero(v > 2)[0]])

# a .* (a > 0.5)  -> zero out small entries via a mask
print("\na * (a > 4):\n", a * (a > 4))

# a(a < 5) = 0  -> in-place conditional assignment (on a copy)
a_copy = a.copy()
a_copy[a_copy < 5] = 0
print("\ncopy with entries < 5 set to 0:\n", a_copy)

print("\nlogical_and / logical_or on masks:")
print(np.logical_and(a > 2, a < 8))
print(np.logical_or(a < 2, a > 8))
```

### 6. Copies, flattening, assignment

```python
y = a.copy()             # MATLAB y = x  (independent copy)
row = a[1, :].copy()     # copy of a single row
flat = a.flatten()       # MATLAB x(:)
print("copied row      :", row)
print("flattened a     :", flat)

z = np.zeros((2, 2))
z[:] = 3                 # MATLAB a(:) = 3
print("z[:] = 3        :\n", z)
```

### 7. Ranges and standard constructors

```python
print("arange(1,11)      :", np.arange(1., 11.))     # MATLAB 1:10
print("arange(10)        :", np.arange(10.))          # MATLAB 0:9
print("column vector     :\n", np.arange(1., 6.)[:, np.newaxis])  # [1:5]'
print("linspace(1,3,4)   :", np.linspace(1, 3, 4))
print("zeros((3,4))      :\n", np.zeros((3, 4)))
print("ones((3,4))       :\n", np.ones((3, 4)))
print("eye(3)            :\n", np.eye(3))
print("diag(a)           :", np.diag(a))              # main diagonal of a
print("diag([1,2,3])     :\n", np.diag([1., 2., 3.]))  # build diagonal matrix
```

```python
rng = np.random.default_rng(0)                       # MATLAB rand(3,4)
print("random 3x4:\n", rng.random((3, 4)))

# meshgrid  (MATLAB [x,y]=meshgrid(...))
xx, yy = np.meshgrid(np.r_[0:4.], np.r_[0:3.])
print("\nmeshgrid X:\n", xx)
print("meshgrid Y:\n", yy)

# repmat -> tile
print("\ntile([[1,2],[3,4]], (2,3)):\n", np.tile(np.array([[1., 2.], [3., 4.]]), (2, 3)))
```

### 8. Reductions: max / min / sum / norm

```python
print("a.max()      (max of all, MATLAB max(max(a))) =", a.max())
print("a.max(0)     (per-column, MATLAB max(a))       =", a.max(0))
print("a.max(1)     (per-row,   MATLAB max(a,[],2))   =", a.max(1))
print("np.maximum(a,b) (element-wise max):\n", np.maximum(a, b))
print("\na.sum()      =", a.sum())
print("a.sum(0)     (column sums) =", a.sum(0))
print("norm(v)      =", np.linalg.norm(v))            # or np.sqrt(v @ v)
```

### 9. Linear algebra (the `scipy.linalg` column)

```python
print("inv(a):\n", linalg.inv(a))
print("\npinv(b):\n", linalg.pinv(b))
print("\nmatrix_rank(a) =", np.linalg.matrix_rank(a))
print("det(a)         =", linalg.det(a))

# a\v  -> solve a x = v
x = linalg.solve(a, v)
print("\nsolve(a, v)    =", x)
print("check a @ x    =", a @ x, " (should equal v =", v, ")")

# least squares  a\v when non-square (use a here for demonstration)
sol, res, rnk, sv = linalg.lstsq(a, v)
print("\nlstsq solution =", sol)
```

```python
# SVD
U, S, Vh = linalg.svd(a)
print("SVD singular values:", S)

# Cholesky  (needs the SPD matrix)
c = linalg.cholesky(spd)
print("\ncholesky(spd) (upper):\n", c)
print("reconstruct c.T @ c:\n", c.T @ c)

# Eigen-decomposition
w, V = linalg.eig(a)
print("\neigenvalues of a:", w)

# QR
Q, R = linalg.qr(a)
print("\nQR - R:\n", R)

# LU
P, L, U = linalg.lu(a)
print("\nLU - L:\n", L)
print("LU - U:\n", U)
```

### 10. FFT, sorting, unique, squeeze

```python
print("fft(v)     :", np.fft.fft(v))
print("ifft(...)  :", np.fft.ifft(np.fft.fft(v)))

print("\nsort(v)    :", np.sort(v))
idx = np.argsort(v)
print("argsort(v) :", idx, " -> sorted:", v[idx])

print("\nunique([3,1,4,1,5,9,2,6,5]):", np.unique(np.array([3,1,4,1,5,9,2,6,5])))

t = np.zeros((1, 3, 1))
print("\nsqueeze shape:", t.shape, "->", t.squeeze().shape)
```
